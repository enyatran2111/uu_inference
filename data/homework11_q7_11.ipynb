## Question 7

```python
import numpy as np
import pandas as pd
from scipy.stats import norm
```

```python
df = pd.read_csv("ames_prices.csv")

# Create indicator: 1 = central air, 0 = no central air
x = (df["Central.Air"] == "Y").astype(int).to_numpy()

# Original sample estimate
p_hat = np.mean(x)
n = len(x)

print("Sample proportion:", p_hat)
```

```python
# Part a: generate 2,000 nonparametric bootrap estimates
seed = 42
R = 2000
rng = np.random.default_rng(seed)

results = []

for t in range(R):
    idx = rng.integers(0, n, n)
    p_hat_t = np.mean(x[idx])
    results.append(p_hat_t)

results = np.array(results)

print("Seed:", seed)
print("R:", R)
```

```python
# Part b: Estimate the standard error with ddof=1
SE = np.std(results, ddof=1)
print("Bootstrap standard error:", SE)
```

```python
# Part c: compute 90%, 95%, and 99% CIs
critical_values = {
    "90%": 1.645,
    "95%": 1.96,
    "99%": 2.576
}

CIs = {}

for level, z_star in critical_values.items():
    lower = p_hat - z_star * SE
    upper = p_hat + z_star * SE

    CIs[level] = (lower, upper)
    print(f"{level} CI: [{lower:.6f}, {upper:.6f}]")
```

```python
# Part d: Hypothesis test for significant level of 0.10, 0.05, 0.01


p0 = 0.90

z = (p_hat - p0) / SE
p_value = 2 * norm.cdf(-abs(z))

print("Z-statistic:", z)
print("Two-sided p-value:", p_value)

for alpha in [0.10, 0.05, 0.01]:
    if p_value < alpha:
        print(f"alpha = {alpha}: Reject H0")
    else:
        print(f"alpha = {alpha}: Fail to reject H0")


```

Part e: All three confidence intervals exclude 0.9 and they are consistent with rejecting the null hypothesis H0 at 0.10, 0.05, and 0.01 significant levels.

| Confidence Level | Confidence Interval | Contains 0.90? | Decision |
|---|---|---|---|
| 90% | [0.9256, 0.9407] | No | Reject H0 |
| 95% | [0.9241, 0.9421] | No | Reject H0 |
| 99% | [0.9213, 0.9449] | No | Reject H0 |

Part f: 
- The estimated proprotion of homes with central air is 93.31% with a standard error of 0.00459.
- The 95% confidence interval is between 92.41% to 94.21%. The p-value is super smal so we can reject the null hypothesis that the true population proportion is 90% at all three significant levels.There is strong statistical evidence that the proportion of homes with central air is greater than 90%.

## Question 8

```python
# Transform area and price
x = np.arcsinh(df["area"].to_numpy())
y = np.arcsinh(df["price"].to_numpy())

# Original Pearson correlation
rho_hat = np.corrcoef(x, y)[0, 1]
n = len(x)

print("Sample size:", n)
print("Estimated correlation:", rho_hat)
```

```python
#  Part a: generate 2,000 nonparametric bootrap estimates


seed = 42
R = 2000
rng = np.random.default_rng(seed)

results = []

for t in range(R):
    idx = rng.integers(0, n, n)
    rho_hat_t = np.corrcoef(x[idx], y[idx])[0, 1]
    results.append(rho_hat_t)

results = np.array(results)

print("Seed:", seed)
print("R:", R)
```

```python
# Part b: Estimate the standard error

SE = np.std(results, ddof=1)
print("Bootstrap standard error:", SE)
```

```python
# Part c: compute 90%, 95%, and 99% CIs


critical_values = {
    "90%": 1.645,
    "95%": 1.96,
    "99%": 2.576
}

CIs = {}

for level, z_star in critical_values.items():
    lower = rho_hat - z_star * SE
    upper = rho_hat + z_star * SE

    CIs[level] = (lower, upper)
    print(f"{level} CI: [{lower:.6f}, {upper:.6f}]")
```

```python
# Part d: Hypothesis test for significant level of 0.10, 0.05, 0.01

rho0 = 0

z = (rho_hat - rho0) / SE
p_value = 2 * norm.sf(abs(z))

print("Z-statistic:", z)
print("Two-sided p-value:", p_value)

for alpha in [0.10, 0.05, 0.01]:
    if p_value < alpha:
        print(f"alpha = {alpha}: Reject H0")
    else:
        print(f"alpha = {alpha}: Fail to reject H0")
```

- All three confidence intervals exclude zero, so we can reject H0 at all three significant levels. Therefore, the estimated correlation of 0.7232 indicates a statistically positive correlation between house area and price. However, this does not establish causation.

Part e:
- Area and price must be resampled together because each pair shows the same house. Resampling separately would break the original relationship between house size and price and this can lead to incorrect bootstrap correlation distribution.

## Question 9

```python
df = pd.read_csv("metabric.csv")

# Include both living and deceased patients
x = df["Overall Survival (Months)"].dropna().to_numpy()

# Original sample estimate
mu_hat = np.mean(x)
n = len(x)

print("Sample size:", n)
print("Estimated mean (months):", mu_hat)
```

```python
# Part a: generate 2,000 nonparametric bootstrap estimates


seed = 42
R = 2000
rng = np.random.default_rng(seed)

results = []

for t in range(R):
    idx = rng.integers(0, n, n)
    mu_hat_t = np.mean(x[idx])
    results.append(mu_hat_t)

results = np.array(results)

print("Seed:", seed)
print("R:", R)
```

```python
# Part b: Compute bootstrap standard error


SE = np.std(results, ddof=1)

print("Bootstrap standard error:", SE)
```

```python
# Part c: compute 90%, 95%, and 99% CIs


critical_values = {
    "90%": 1.645,
    "95%": 1.96,
    "99%": 2.576
}

CIs = {}

for level, z_star in critical_values.items():
    lower = mu_hat - z_star * SE
    upper = mu_hat + z_star * SE

    CIs[level] = (lower, upper)
    print(f"{level} CI: [{lower:.4f}, {upper:.4f}] months")
```

```python
# Part d: Hypothesis test for significant level of 0.10, 0.05, 0.01


mu0 = 120

z = (mu_hat - mu0) / SE
p_value = 2 * norm.sf(abs(z))

print("Z-statistic:", z)
print("Two-sided p-value:", p_value)

for alpha in [0.10, 0.05, 0.01]:
    if p_value < alpha:
        print(f"alpha = {alpha}: Reject H0")
    else:
        print(f"alpha = {alpha}: Fail to reject H0")
```

Part e:
- The estimated average time to death or last follow-up is 129.09 months (10.76 years). The 90%, 95%, and 99% confidence intervals are [125.58, 132.61], [124.91, 133.28], and [123.59, 134.60] months, respectively, equivalent to [10.47, 11.05], [10.41, 11.11], and [10.30, 11.22] years.
- All three confidence intervals exclude 120 months so we can reject the null hypothesis at 10%, 5%, and 1% significant level. There is statistically significant evidence that the mean team exceeds 120 months. 
- The results do not represent actual average lifetime because living patients have not died yet. For these patients, we only know how long they were followed, so we only estimate average recorded time rather than actual lifetime.

```python
# Part a: generate 2,000 nonparametric bootrap estimates


seed = 42
R = 2000
rng = np.random.default_rng(seed)

results = []

for t in range(R):
    idx = rng.integers(0, n, n)
    mu_hat_t = np.mean(x[idx])
    results.append(mu_hat_t)

results = np.array(results)

print("Seed:", seed)
print("R:", R)
```

```python
# Part b: Estimate the standard error
SE = np.std(results, ddof=1)

print("Bootstrap standard error:", SE)
```

```python
# Part c: compute 90%, 95%, and 99% CIs


critical_values = {
    "90%": 1.645,
    "95%": 1.96,
    "99%": 2.576
}

CIs = {}

for level, z_star in critical_values.items():
    lower = mu_hat - z_star * SE
    upper = mu_hat + z_star * SE

    CIs[level] = (lower, upper)
    print(f"{level} CI: [{lower:.4f}, {upper:.4f}] hours")
```

```python
# Part d: Hypothesis test for significant level of 0.10, 0.05, 0.01


mu0 = 8

z = (mu_hat - mu0) / SE
p_value = 2 * norm.sf(abs(z))

print("Z-statistic:", z)
print("Two-sided p-value:", p_value)

for alpha in [0.10, 0.05, 0.01]:
    if p_value < alpha:
        print(f"alpha = {alpha}: Reject H0")
    else:
        print(f"alpha = {alpha}: Fail to reject H0")
```

Part e:
- The estimated mean weekday sleep is 7.62 hours, about 23 minutes below 8 hours. Because all three confidence intervals exclude 8 hours, we reject the null hypothesis at the 10%, 5%, and 1% siginificant levels. There is no statistically significant evidence that the population average weekday sleep is 8 hours.
- Although the result is statistically significant, 23 minutes less sleep may not be practically meaningful. Because we ignored the survey design, we treated everone as independently sampled and equally weighted. Howwever, NHANES data used a more complex sampling method to account for clustering and stratification so our standard error and confidence interval may be inaccurate.
